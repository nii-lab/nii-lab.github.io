<span id="requests-정상-설치-확인"></span>

### Verify the Requests installation

```python
import requests               

url = 'https://news.daum.net/'  
response = requests.get(url)    
html_text = response.text
print(html_text[:500])
```

<span id="beautifulsoup-정상-설치-확인"></span>

### Verify the Beautifulsoup installation

```python
from bs4 import BeautifulSoup as bs

html = bs(html_text, 'html.parser')
print(html.select('link'))
```

### select()

```python
html_hyperlink_list = html.select('a')
sample = html_hyperlink_list[5]
print(sample)
print(sample.text)
print(sample['class'])
```

<span id="tag-내의-tag-찾기"></span>

### Find a tag inside a tag

```python
html_list = html.select('link script')
print(html_list[0])
```

<span id="class로-tag-찾기"></span>

### Find tags by class

```python
html_list_by_class = html.select('.item_issue')
print(html_list_by_class[0])
```

<span id="if로-tag-찾기"></span>

### Find tags using if

```python
html_list_by_id = html.select('#mainContent')
print(html_list_by_id[0])
```

<span id="뉴스-헤드라인-패턴-찾기"></span>

### Find news headline patterns

```python
headlines = html.select('.item_issue .tit_g a')
print(headlines[0])
```

<span id="tag에서-text-분리하기"></span>

### Extract text from a tag

```python
for headline in headlines:
  print(headline.text)
```

<span id="text에서-불필요한-줄바꿈-없애기"></span>

### Remove unnecessary line breaks from text

```python
for headline in headlines:
  print(headline.text.replace('\n', ''))
```

<span id="text에서-추가로-불필요한-띄어쓰기-없애기"></span>

### Remove additional unnecessary spaces from text

```python
for headline in headlines:
  print(headline.text.replace('\n', '').replace('  ', ''))
```

\###Complete news headline scraping with numbered headlines

```python
headlines = html.select('.item_issue .tit_g a')
for i, headline in enumerate(headlines):
  headline_text = headline.text.replace('\n', '').replace('  ', '')
  print(i, headline_text)
```

<span id="기상청-날씨-스크랩"></span>

### Scrape Korea Meteorological Administration weather data

```python
url = 'https://www.weather.go.kr/w/obs-climate/land/city-obs.do'  
response = requests.get(url)    
html_text = response.text
html = bs(html_text, 'html.parser')
cities = html.select('tbody tr')

temperatures = {}
for city in cities:
  city_name = city.select('a')[0].text
  temp = city.select('td')[5].text
  temperatures[city_name] = temp
print(temperatures)
```
